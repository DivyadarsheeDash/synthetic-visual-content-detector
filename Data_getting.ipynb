```python
train_cont = []
```

```python
for root , sub , file in os.walk(train_add):
    print(f"folder is {root}")
    print(f"sub folder is {sub}")
    for i in file:
        if len(sub) == 0:
            train_cont.append(os.path.join(root , i))
```

```python
train_container = {}
```

```python
for i in train_cont:
    if 'FAKE' in i:
        train_container[i] = 'FAKE'
    elif 'REAL' in i:
        train_container[i] = 'REAL'
```

```python
len(train_container)
```

```python
import pandas as pd
```

```python
train_df = pd.DataFrame( train_container.items(), columns = ["image_add", "Class"])
```

```python
train_df["Class"].value_counts()
```

```python
train_df.to_csv("Train_Data.csv" , index = False)
```
